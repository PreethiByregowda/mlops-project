```python
from datasets import load_dataset
import pandas as pd

# Load GoEmotions
ds = load_dataset("go_emotions")

# Get label names
label_names = ds["train"].features["labels"].feature.names

# Take first 50 rows and convert to DataFrame
df = pd.DataFrame(ds["train"][:50])

# Add empty columns for each label
for label in label_names:
    df[label] = 0

# Fill emotion columns with 1 where applicable
def fill_emotions(row):
    for idx in row["labels"]:
        label = label_names[idx]
        row[label] = 1
    return row

df = df.apply(fill_emotions, axis=1)

# Extract emotion names per row
def get_emotions(row):
    return [emotion for emotion in label_names if row[emotion] == 1]

df["emotion_names"] = df.apply(get_emotions, axis=1)

# Print just text + emotions nicely
for i, row in df.iterrows():
    print(f"Text: {row['text']}")
    print(f"Emotions: {row['emotion_names']}")
    print("---")
```
